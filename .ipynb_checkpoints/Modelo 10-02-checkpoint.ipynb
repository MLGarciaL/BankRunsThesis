```python
DF10 = df_results[(df_results['N'] == 10) & (df_results['lambda'] == 0.3) & (df_results['threshold'] == None)]
DF11 = df_results[(df_results['N'] == 10) & (df_results['lambda'] == 0.3) & (df_results['threshold'] == 0.1)]
DF12 = df_results[(df_results['N'] == 10) & (df_results['lambda'] == 0.3) & (df_results['threshold'] == 0.2)]
DF13 = df_results[(df_results['N'] == 10) & (df_results['lambda'] == 0.3) & (df_results['threshold'] == 0.3)]
DF14 = df_results[(df_results['N'] == 10) & (df_results['lambda'] == 0.3) & (df_results['threshold'] == 0.4)]
DF15 = df_results[(df_results['N'] == 10) & (df_results['lambda'] == 0.3) & (df_results['threshold'] == 0.5)]
DF16 = df_results[(df_results['N'] == 10) & (df_results['lambda'] == 0.3) & (df_results['threshold'] == 0.6)]

table10 = (DF10.groupby('p', as_index=False)['bank_failure'].mean().rename(columns={'bank_failure': 'average_bank_failure'}))
table11 = (DF11.groupby('p', as_index=False)['bank_failure'].mean().rename(columns={'bank_failure': 'average_bank_failure'}))
table12 = (DF12.groupby('p', as_index=False)['bank_failure'].mean().rename(columns={'bank_failure': 'average_bank_failure'}))
table13 = (DF13.groupby('p', as_index=False)['bank_failure'].mean().rename(columns={'bank_failure': 'average_bank_failure'}))
table14 = (DF14.groupby('p', as_index=False)['bank_failure'].mean().rename(columns={'bank_failure': 'average_bank_failure'}))
table15 = (DF15.groupby('p', as_index=False)['bank_failure'].mean().rename(columns={'bank_failure': 'average_bank_failure'}))
table16 = (DF16.groupby('p', as_index=False)['bank_failure'].mean().rename(columns={'bank_failure': 'average_bank_failure'}))
```

```python
import matplotlib.pyplot as plt

plt.plot(table10['p'],table10['average_bank_failure'], '-o', label  = 'threshold = None')
plt.plot(table11['p'],table11['average_bank_failure'], '-o', label  = 'threshold = 0.1')
plt.plot(table12['p'],table12['average_bank_failure'], '-o', label  = 'threshold = 0.2')
plt.plot(table13['p'],table13['average_bank_failure'], '-o', label  = 'threshold = 0.3')
plt.plot(table14['p'],table14['average_bank_failure'], '-o', label  = 'threshold = 0.4')
plt.plot(table15['p'],table15['average_bank_failure'], '-o', label  = 'threshold = 0.5')
plt.plot(table16['p'],table16['average_bank_failure'], '-o', label  = 'threshold = 0.6')
plt.xlabel('p')
plt.ylabel('Average bank failure')
plt.title(r'Average Bank Failure by p, $\lambda = 0.3$')
plt.grid(ls = 'dashed', lw = 0.5)
plt.ylim(0,1.05)
plt.legend(loc ='upper left')
plt.show()
```
