```python
import datetime
from scipy.interpolate import make_interp_spline



def plot_graph(df, type_network, N, Lambda, show = True, save = True):
    # Filter the DataFrame based on the provided parameters
    df_filtered = df[(df['N'] == N) & (df['lambda'] == Lambda)]

    if N == 0:
        N = 100

    if type_network == 'CLUSTERED':
        for cluster in df_filtered['cluster'].unique():
            df_cluster = df_filtered[df_filtered['cluster'] == cluster]
            subtitle = rf'Model Parameters {type_network} (Cluster {cluster}): $N ={N}$, $\lambda = {Lambda}$'
            name = f'outputs/SIMULATIONS/{type_network}/{N}/{cluster}/RUN_DENSITY_Network{type_network}_Cluster{cluster}_N{N}_Lambda{Lambda}.png'
            plot_run_percentage_vs_density(df_cluster, Lambda, name, subtitle, show=show, save=save)
    elif type_network == 'WS':
        for p in df_filtered['p'].unique():
            df_p = df_filtered[df_filtered['p'] == p]
            subtitle = rf'Model Parameters {type_network} (p = {p}): $N ={N}$, $\lambda = {Lambda}$'
            name = f'outputs/SIMULATIONS/{type_network}/{N}/{p}/RUN_DENSITY_Network{type_network}_P{p}_N{N}_Lambda{Lambda}.png'
            plot_run_percentage_vs_density(df_p, Lambda, name, subtitle, show=show, save=save)
    
    
    subtitle =rf'Model Parameters {type_network}: $N ={N}$, $\lambda = {Lambda}$'
    name = f'outputs/SIMULATIONS/{type_network}/{N}/RUN_DENSITY_Network{type_network}_N{N}_Lambda{Lambda}.png'

    plot_run_percentage_vs_density(df_filtered, Lambda, name, subtitle, show=show, save=save)

def plot_run_percentage_vs_density(df_filtered, Lambda, name, subtitle, show=True, save=True):
    
    # Aggregate the data by density_bin and calculate the mean run_percentage
    agg = df_filtered.groupby('density_bin10')['run_percentage'].mean().reset_index().sort_values('density_bin10')

    if df_filtered['density_bin10'].min() > 0:
        # Add a point at density 0 with a run percentage of Lambda
        agg.loc[len(agg)] = [0.0, Lambda]
        agg = agg.sort_values('density_bin10').reset_index(drop=True)

    # Extract raw coordinates
    x_raw = agg['density_bin10'].values
    y_raw = agg['run_percentage'].values
    
    # Smooth trendline using B-Spline kernel smoothing
    x_smooth = np.linspace(x_raw.min(), x_raw.max(), 300)
    if len(x_raw) >= 4:
        spline = make_interp_spline(x_raw, y_raw, k=3)
        y_smooth = spline(x_smooth)
    else:
        # Not enough points for a cubic spline
        y_smooth = np.interp(x_smooth, x_raw, y_raw)

    y_smooth = np.clip(y_smooth, Lambda, 1)
        
    # Plotting
    plt.figure(figsize=(9, 5.5), dpi=300)


    # Plot smoothed kernel regression line
    plt.plot(x_smooth, y_smooth, color='#4A90E2', linewidth=2.5, label='Kernel Smoothing', zorder=2)
    
    # Plot raw scatter data points
    plt.scatter(x_raw, y_raw, color='#4A90E2', edgecolor='#2C3E50', linewidth=1.2, s=25, zorder=3, label='Binned Means')
    plt.scatter(df_filtered['density_bin'], df_filtered['run_percentage'], color='#555555', linewidth=1.2, s=5, zorder=1, alpha = 0.05, label='Raw Data Points')
    
    # Set axis limits and ticks
    plt.xlim(-0.02, 1.02)
    plt.xticks(np.arange(0, 1.1, 0.1))
    plt.ylim(0, 1.02)
    plt.xlabel('Network Density', fontsize=12, labelpad=10)
    plt.ylabel('Average Share of Depositors Running', fontsize=12, labelpad=10)
    plt.grid(ls = 'dashed', lw = 0.5)

    # Title & Subtitle structure matching your thesis standard
    plt.title('Bank Run Dynamics Across Network Density', loc='left', fontweight='bold', pad=25, fontsize=16)
    plt.text(0,1.02,subtitle,transform=plt.gca().transAxes,fontsize=12,ha='left',va='bottom',color='#555555')

    # Remove top and right spines for a clean publication look
    plt.gca().spines['top'].set_visible(False)
    plt.gca().spines['right'].set_visible(False)
    plt.gca().spines['left'].set_color('#BDC3C7')
    plt.gca().spines['bottom'].set_color('#BDC3C7')

    plt.legend(loc="center left", bbox_to_anchor=(1.03, 0.5),  facecolor="#F8F9F9", edgecolor="#BDC3C7", fontsize=10, borderpad=0.8, labelspacing=1.0)

    plt.tight_layout()

    # Save options
    if save:
        plt.savefig(name, dpi=300, bbox_inches='tight', transparent=True)
    if show:
        plt.show()

    plt.close()
```
